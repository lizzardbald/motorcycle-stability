### 3.2 Gyroscopic Precession

Every motorcycle rider takes it for granted -- lean the bike into a corner at 100 km/h and it holds its line. Lean the same bike at walking pace and it topples. The difference is **gyroscopic angular momentum**: the spinning wheels generate a force that resists changes to their orientation, and at speed this force is strong enough to keep the machine stable through extreme lean angles.

This same force, however, has a darker side. The very mechanism that lets you drag a knee at 55 degrees of lean is also a key ingredient in the tank slapper.

### Why Low Trail Invites the Death Wobble

Trail is the front end's restoring lever arm. When the wheel gets knocked off-center -- by a bump, a gust of wind, or a rough landing -- the lateral force at the contact patch acts through the trail distance to create a **restoring torque** that steers the wheel back into line:

$$
T_{restore} = F_{lateral} \times t
$$

With generous trail, this restoring torque is strong and decisive -- the wheel snaps back to center and the disturbance dies quickly. But as trail decreases, two things happen simultaneously:

1. **The restoring torque weakens.** A shorter lever arm means less corrective force for the same disturbance. The wheel still tries to self-center, but it does so with less authority -- and less authority means it takes longer to settle, giving the oscillation more time to develop.

2. **The overshoot grows.** When the wheel does correct, it carries momentum past the center line. With strong trail, the restoring torque catches the overshoot early and damps it. With weak trail, the wheel swings further past center before the geometry can pull it back -- and each swing feeds energy into the next.

This is the mechanism behind the wobble mode. The front end oscillates back and forth around the steering axis, and if the damping in the system (friction in the bearings, tire hysteresis, steering damper) is insufficient to absorb the energy being fed into each cycle, the amplitude grows.

The critical insight: **trail does not just set the magnitude of the restoring force -- it sets the frequency and damping characteristics of the entire steering oscillation.** Lower trail raises the natural frequency of the wobble mode and reduces its damping ratio. At a certain forward speed, that damping ratio crosses zero, and the oscillation becomes self-sustaining.

This is why sportbikes -- with their 90mm trail -- are the most common victims. They live closer to the edge by design.
